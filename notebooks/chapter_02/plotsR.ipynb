```python
def plot_houwie_summary(data: pd.DataFrame,
                        trait: str,
                        width: int = 10,
                        height: int = 10,
                        meanprops_cd: dict[str, str] = {"marker": 'o', "markerfacecolor": "red", "markeredgecolor": "black"},
                        meanprops_cid: dict[str, str] = {"marker": 'o', "markerfacecolor": "blue", "markeredgecolor": "black"},
                        flierprops_cd = {'marker': 'o', 'markerfacecolor': "none", "markeredgecolor": "red"}, # for outliers
                        flierprops_cid = {'marker': 'o', 'markerfacecolor': "none", "markeredgecolor": "blue"},
                        medianprops_cd: dict[str, str]|None = None,
                        medianprops_cid: dict[str, str]|None = None
                       ) -> None:
    
    METRICS_AND_LATEX_SYMBOLS = { "rates": "Rates", "alpha": r"$\alpha$", "sigma.sq": r"$\sigma^{2}$", "theta": r"$\theta$" } # LaTex symbols for some of the model estimates
    fig, axes = plt.subplots(nrows=2, ncols=2, sharex=True, sharey=False)
    fig.set_size_inches(width, height)

    _cd = data.query(f"trait=='{trait}' and cd==True")
    _cid = data.query(f"trait=='{trait}' and cd==False")
    
    for (axis, col) in zip(axes.flatten(), METRICS_AND_LATEX_SYMBOLS):
    
        axis.boxplot([df.loc[:, col] for (_, df) in _cd.groupby("tip_state")], showmeans=True, meanprops=meanprops_cd, medianprops=medianprops_cd, flierprops=flierprops_cd)
        axis.boxplot([df.loc[:, col] for (_, df) in _cid.groupby("tip_state")], showmeans=True, meanprops= meanprops_cid, medianprops=medianprops_cid, flierprops=flierprops_cid)
    
        axis.set_xticks(ticks=np.arange(1, data.tip_state.unique().size + 1), labels=[_ for (_, df) in _cd.groupby("tip_state")]) # set it last manually, instead of through axes.boxplot() each time
        axis.set_ylabel(METRICS_AND_LATEX_SYMBOLS.get(col))
        axis.grid()
        axis.set_axisbelow(True)

    fig.supxlabel(f"${trait}$")
    fig.legend(edgecolor="black", ncols=2, loc="upper right", handles=[Line2D(xdata=[0], ydata=[0], marker='o', markeredgecolor="black", markerfacecolor="red", label="CD", linestyle="none"),
                                                             Line2D(xdata=[0], ydata=[0], marker='o', markeredgecolor="black", markerfacecolor="blue", label="CID", linestyle="none")])
    plt.tight_layout()
    
```

```python
plot_houwie_summary(data=summary, trait="RD")
plt.show()

# plt.savefig(r"../../plots/houwie_summary_rd_995sp_100sims.png", format="png", dpi=800, transparent=True)
```

```python
def plot_houwie_summary_v2(data: pd.DataFrame, width: int = 16, height: int = 16, ccd: str = "red", ccid: str = "blue") -> None:

    TRAITS = ("RD", "SRL", "RTD")
    METRICS_AND_LATEX_SYMBOLS = { "alpha": r"$\alpha$", "sigma.sq": r"$\sigma^{2}$", "theta": r"$\theta$" }
    fig, axes = plt.subplots(nrows=3, ncols=3, sharex=True, sharey=False)
    fig.subplots_adjust(wspace=0.25, hspace=0.1)
    fig.set_size_inches(width, height)

    for (trait, row) in zip(TRAITS, axes):
        
        subset_cd = data.query(f"trait=='{trait}' and cd==True")
        subset_cid = data.query(f"trait=='{trait}' and cd==False")
        axis_column = 0 # nth column of the current row being plotted

        for (axis, est) in zip(row, METRICS_AND_LATEX_SYMBOLS):
            axis_column += 1
            axis.scatter(x=subset_cd.tip_state, y=subset_cd.rates, ec=ccd, fc="none")
            axis.errorbar(
                x=subset_cd.groupby("tip_state", as_index=False).agg({est: "mean"})["tip_state"], # states as thex axis
                y=subset_cd.groupby("tip_state", as_index=False).agg({est: "mean"})[est], # mean of the metric as the y axis
                yerr=subset_cd.groupby("tip_state", as_index=False).agg({est: sem})[est], # standard error of the metric for error bars
                fmt='o', ecolor="black", capsize=7.5, elinestyle="-", mfc=ccd, mec="black"
            )
    
            axis.scatter(x=subset_cid.tip_state, y=subset_cid.rates, ec=ccid, fc="none")
            axis.errorbar(
                x=subset_cid.groupby("tip_state", as_index=False).agg({est: "mean"})["tip_state"],
                y=subset_cid.groupby("tip_state", as_index=False).agg({est: "mean"})[est],
                yerr=subset_cid.groupby("tip_state", as_index=False).agg({est: sem})[est], fmt='o', ecolor="black", capsize=7.5, elinestyle="-", mfc=ccid, mec="black"
            )
            
            
            if trait == "RD":
                axis.set_title(METRICS_AND_LATEX_SYMBOLS.get(est)) # set the title only for the first row
            if axis_column == 1:
                axis.set_ylabel(f"${trait}$") # set the trait name as the y axis label for the first plot in the row
            axis.grid()
            axis.set_axisbelow(True)

    # fig.supylabel(f"${trait}$")
    # fig.legend(edgecolor="black", ncols=2, loc="lower right", handles=[Line2D(xdata=[0], ydata=[0], marker='o', markeredgecolor="black", markerfacecolor="red", label="CD", linestyle="none"),
    #            Line2D(xdata=[0], ydata=[0], marker='o', markeredgecolor="black", markerfacecolor="blue", label="CID", linestyle="none")])
    # plt.tight_layout()
    
```

```python
plot_houwie_summary_v2(data=data)
plt.tight_layout()
plt.savefig(r"../../plots/houwie1301_250sims.png", format="png", dpi=1000, transparent=False)
```

```python
df = pd.read_csv(r"../../data/chapter2/hOUwie-update_getModelAvgParams.csv")
df.head()
```

```python
fig, (sigsq_axes, theta_axes) = plt.subplots(nrows=2, ncols=3, sharex=True)
fig.set_size_inches(12, 8)

# look at parameters sigma.sq and theta of the CID fits

for (i, (trait, sub)) in enumerate(df.query("cd==False").groupby("trait", as_index=False)):
    avg = sub.groupby("tip_state").aggregate({"sigma.sq": "mean", "theta": "mean"}) # means
    stderr = sub.groupby("tip_state").aggregate({"sigma.sq": sem, "theta": sem}) # standard errors
    
    # sigsq_axes[i].plot(avg["sigma.sq"], linestyle = "none", marker = 'o', mfc = "blue", mec = "black")
    sigsq_axes[i].errorbar(x = avg["sigma.sq"].index, y = avg["sigma.sq"].values,
                           yerr = stderr["sigma.sq"].values,
                           linestyle = "none", ecolor = "black", capsize = 4, marker = 'o', mfc = "blue", mec = "black")
    sigsq_axes[i].set_title(trait)
    sigsq_axes[i].grid()
    sigsq_axes[i].set_axisbelow(True)

    # theta_axes[i].plot(avg["theta"], linestyle = "none", marker = 'o', mfc = "red", mec = "black")
    theta_axes[i].errorbar(x = avg["theta"].index, y = avg["theta"].values, yerr = stderr["theta"].values,
                           linestyle = "none", ecolor = "black", capsize = 4, marker = 'o', mfc = "red", mec = "black")
    theta_axes[i].grid()
    theta_axes[i].set_axisbelow(True)

plt.tight_layout()
plt.show()
```
