```r
suppressPackageStartupMessages({
    library("ape")
    library("phytools")
    library("phylolm")
    library("effects")
    library("ggplot2")
    library("lme4")
})
```

```r
STATE_COLOURS <- c(AM = "#E7BC0C", EcM = "#23B9AF", ErM = "#6B6CF2", NM = "#11E823", NMAM = "#2D2E30", EcMAM = "#E71A23")
```

```r
tree <- ape::read.tree("../../data/chapter2/uphylomaker/FRED4_1301.tre")
if(!ape::is.binary(tree)) tree <- ape::multi2di(tree, random = FALSE)

tree$edge.length[tree$edge.length <= 0] <- rnorm(n = sum(tree$edge.length == 0), mean = 1e-6, sd = 1e-8)
stopifnot(!sum(tree$edge.length <= 0))

if (!ape::is.ultrametric(tree)) tree <- phytools::force.ultrametric(tree = tree, method = "extend")
stopifnot(ape::is.binary(tree) && ape::is.ultrametric(tree))

traits <- read.csv("../../data/chapter2/FRED/subsets/name_matched_FRED4_1301.csv")[, c("binominal", "F00679", "F00727", "F00709", "state")]
stopifnot(all(tree$tip.label==traits$binominal))
```
